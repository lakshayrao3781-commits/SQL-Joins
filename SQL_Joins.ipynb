Question 1. Retrieve all customers who have placed at least one order.

SELECT DISTINCT c.*
FROM Customers c
INNER JOIN Orders o ON c.CustomerID = o.CustomerID;

Result: John Smith, Mary Johnson, Peter Adams.

Question 2. Retrieve all customers and their orders, including customers who have not placed any orders.

SELECT c.CustomerID, c.CustomerName, o.OrderID, o.OrderDate, o.Amount
FROM Customers c
LEFT JOIN Orders o ON c.CustomerID = o.CustomerID;

Result: Includes all 5 customers (Nancy Miller and Robert White will show NULL for order details).

Question 3. Retrieve all orders and their corresponding customers, including orders placed by unknown
customers.

SELECT o.OrderID, o.Amount, c.CustomerID, c.CustomerName
FROM Customers c
RIGHT JOIN Orders o ON c.CustomerID = o.CustomerID;

Result: Orders 101, 102, 103, 104, and Order 105 (placed by CustomerID 6, showing NULL for customer name).

Question 4. Display all customers and orders, whether matched or not.

-- MySQL does not support FULL OUTER JOIN directly, so we use UNION
SELECT c.CustomerID, c.CustomerName, o.OrderID, o.Amount
FROM Customers c
LEFT JOIN Orders o ON c.CustomerID = o.CustomerID
UNION
SELECT c.CustomerID, c.CustomerName, o.OrderID, o.Amount
FROM Customers c
RIGHT JOIN Orders o ON c.CustomerID = o.CustomerID;

Result: Displays all customers (including those without orders) and all orders (including order 105 without a matching customer).

Question 5. Find customers who have not placed any orders.

SELECT c.*
FROM Customers c
LEFT JOIN Orders o ON c.CustomerID = o.CustomerID
WHERE o.OrderID IS NULL;

Result: Nancy Miller (CustomerID: 4) and Robert White (CustomerID: 5)

Question 6. Retrieve customers who made payments but did not place any orders.

SELECT DISTINCT c.*
FROM Customers c
INNER JOIN Payments p ON c.CustomerID = p.CustomerID
LEFT JOIN Orders o ON c.CustomerID = o.CustomerID
WHERE o.OrderID IS NULL;

Result: Nancy Miller (CustomerID: 4 — made payment P004 but has no orders).

Question 7. Generate a list of all possible combinations between Customers and Orders.

SELECT c.CustomerID, c.CustomerName, o.OrderID, o.Amount
FROM Customers c
CROSS JOIN Orders o;

Result: 25 rows total (5 customers × 5 orders).

Question 8. Show all customers along with order and payment amounts in one table.

SELECT
    c.CustomerID,
    c.CustomerName,
    o.Amount AS OrderAmount,
    p.Amount AS PaymentAmount
FROM Customers c
LEFT JOIN Orders o ON c.CustomerID = o.CustomerID
LEFT JOIN Payments p ON c.CustomerID = p.CustomerID;

Result: Returns customer records joined with their order and payment amounts.

Question 9. Retrieve all customers who have both placed orders and made payments.

SELECT DISTINCT c.*
FROM Customers c
INNER JOIN Orders o ON c.CustomerID = o.CustomerID
INNER JOIN Payments p ON c.CustomerID = p.CustomerID;

Result: John Smith, Mary Johnson, Peter Adams (CustomerIDs: 1, 2, 3).
